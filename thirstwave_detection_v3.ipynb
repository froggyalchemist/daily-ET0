```python
et0_rad_file = f"/work10/archive/CMIP6/CMIP-SSPs/outputs/{model}/{exp}/{model}_{exp}_daily_ET0rad_18500101-20141231.nc"
et0_adv_file = f"/work10/archive/CMIP6/CMIP-SSPs/outputs/{model}/{exp}/{model}_{exp}_daily_ET0adv_18500101-20141231.nc"

rad = xr.open_dataset(et0_rad_file, chunks={"time": 365})
adv = xr.open_dataset(et0_adv_file, chunks={"time": 365})

# Drop Feb. 29th
rad = rad.convert_calendar('noleap', align_on='year')
adv = adv.convert_calendar('noleap', align_on='year')

# Baseline: 1981–2000
rad_cli = rad.sel(time=slice('1981', '2000')).mean(dim='time')["ET0rad"]
adv_cli = adv.sel(time=slice('1981', '2000')).mean(dim='time')["ET0adv"]
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 10))

BINS = np.arange(0, 6.5, step=0.5)
rad_cli.plot(ax=axes[0, 0])
xr.plot.hist(rad_cli, ax=axes[0, 1], bins=BINS)

adv_cli.plot(ax=axes[1, 0])
xr.plot.hist(adv_cli, ax=axes[1, 1], bins=BINS)

XLIM = (0, 6)
YLIM = (0, 14000)
for ax in axes[:, 1]:
    ax.set_xlim(*XLIM)
    ax.set_ylim(*YLIM)

plt.tight_layout()
```
