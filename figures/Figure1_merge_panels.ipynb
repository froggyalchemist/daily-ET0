```python
from PIL import Image
import matplotlib.pyplot as plt

# ============================================================
# Read images
# ============================================================
img_top   = Image.open("png/Figure1_upper.png")  # <- detection figure
img_left  = Image.open("png/Figure1_left.png")
img_right = Image.open("png/Figure1_right.png")

# ============================================================
# Create layout
# ============================================================
fig = plt.figure(figsize=(18, 19.5))

gs = fig.add_gridspec(
    2, 2,
    height_ratios=[4.25, 15],   # top panel smaller
    width_ratios=[1, 1],
    hspace=-0.13,
    wspace=-0.02
)

# ============================================================
# Top panel
# ============================================================
ax_top = fig.add_subplot(gs[0, :])

ax_top.imshow(img_top)
ax_top.axis("off")
pos = ax_top.get_position()
pos.height
ax_top.set_position([
    pos.x0 - 0.02,   # 負值往左；正值往右
    pos.y0,
    pos.width,
    pos.height
])
# ============================================================
# Bottom left
# ============================================================
ax_left = fig.add_subplot(gs[1, 0])

ax_left.imshow(img_left)
ax_left.axis("off")

# ============================================================
# Bottom right
# ============================================================
ax_right = fig.add_subplot(gs[1, 1])

ax_right.imshow(img_right)
ax_right.axis("off")

# ============================================================
# Save
# ============================================================

for ext in ["png", "pdf"]:
    filename = f"{ext}/Figure1_full.{ext}"

    fig.savefig(
        filename,
        dpi=500,
        bbox_inches="tight",
        pad_inches=0.08
    )

    print(f"✓ {filename}")

plt.show()
```
