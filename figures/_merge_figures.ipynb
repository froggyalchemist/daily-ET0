```python
from PIL import Image

# Left column: A, C, E, G (time series)
img_left = Image.open("annual_thirstwave_metrics_growing_season.png").convert("RGBA")
# Right column: B, D, F, H (maps)
img_right = Image.open("end_of_century_anom.png").convert("RGBA")

w1, h1 = img_left.size
w2, h2 = img_right.size
print("left:", img_left.size, "right:", img_right.size)

# Canvas sized to fit both, top-aligned, white background
combined = Image.new("RGBA", (w1 + w2, max(h1, h2)), (255, 255, 255, 255))
combined.paste(img_left, (0, 0), img_left)
combined.paste(img_right, (w1, 0), img_right)

combined.convert("RGB").save("Figure3.png", dpi=(500, 500))
```
